# Atividade 5 - IDH por Unidade da Federação
Evolução do IDH de 1991 a 2024 com pandas (`melt`) e matplotlib.

````python
import pandas as pd
import matplotlib.pyplot as plt
````

## 1. Abrir o CSV com `;` e vírgula decimal
A primeira linha do arquivo é apenas um título, então pulamos com `skiprows=1`. O texto está em `latin1` e os números decimais usam vírgula (`0,402`), por isso usamos `decimal=","`.

````python
df = pd.read_csv(
    r'ATIVIDADE 5 MELT MATPLOTLIB/Tabela4.csv',
    sep=';',
    decimal=',',
    encoding='latin1',
    skiprows=1,
)

df
````

## 2. Limpar colunas vazias
Procuramos colunas totalmente vazias (todas as linhas são NaN) e as removemos. Neste arquivo todas as colunas estão preenchidas.

````python
colunas_vazias = [c for c in df.columns if df[c].isna().all()]
print('Colunas vazias encontradas:', colunas_vazias)

if colunas_vazias:
    df = df.drop(columns=colunas_vazias)

df
````

## 3. Ordenar os estados por maior IDH em 2024

````python
df_2024 = df.sort_values('2024', ascending=False).reset_index(drop=True)

df_2024[['Sigla', 'Estado', '1991', '2024']].head(10)
````

## 4. Estado com a maior melhora de IDH entre 2024 e 1991

````python
df['DIFERENCA_2024_1991'] = df['2024'] - df['1991']

melhor_estado = df.sort_values('DIFERENCA_2024_1991', ascending=False).iloc[0]
print(f"Maior melhora: {melhor_estado['Estado']} "
      f"({melhor_estado['Sigla']}) com +{melhor_estado['DIFERENCA_2024_1991']:.3f} "
      f"de IDH, indo de {melhor_estado['1991']:.3f} para {melhor_estado['2024']:.3f}")
````

## 5. Existe algum estado em que o IDH piorou?

````python
piorou = df[df['DIFERENCA_2024_1991'] < 0]

if piorou.empty:
    print('Nenhum estado piorou o IDH entre 1991 e 2024.')
else:
    piorou[['Sigla', 'Estado', '1991', '2024', 'DIFERENCA_2024_1991']]
````

## 6. Passar a tabela do formato largo para o formato longo (`melt`)

Sintaxe do `melt`:
```python
anos = [c for c in df.columns if str(c).isdigit()]
id_vars = [c for c in df.columns if c not in anos]
df_longo = df.melt(id_vars=id_vars, value_vars=anos, var_name='Ano', value_name='IDH')
df_longo['Ano'] = df_longo['Ano'].astype(int)
df_longo['IDH'] = pd.to_numeric(df_longo['IDH'], errors='coerce')
```

````python
anos = [c for c in df.columns if str(c).isdigit()]
print(anos)

id_vars = [c for c in df.columns if c not in anos]
df_longo = df.melt(
    id_vars=id_vars,
    value_vars=anos,
    var_name='Ano',
    value_name='IDH',
)
df_longo['Ano'] = df_longo['Ano'].astype(int)
df_longo['IDH'] = pd.to_numeric(df_longo['IDH'], errors='coerce')

df_longo
````

## 7. Plotar apenas Minas Gerais

````python
%matplotlib inline

fig, ax = plt.subplots(figsize=(9, 5))

mg = df_longo[df_longo['Sigla'] == 'MG'].sort_values('Ano')
ax.plot(mg['Ano'], mg['IDH'], marker='o')
ax.set_title('Evolução do IDH - Minas Gerais (1991-2024)')
ax.set_xlabel('Ano')
ax.set_ylabel('IDH')
ax.grid(True, linestyle='--', alpha=0.5)
fig.tight_layout()
plt.show()
````

## 8. Plotar a evolução do IDH de cada estado

````python
fig, ax = plt.subplots(figsize=(12, 7))

for sigla, grupo in df_longo.groupby('Sigla'):
    grupo = grupo.sort_values('Ano')
    ax.plot(grupo['Ano'], grupo['IDH'], marker='o', markersize=3, linewidth=1.5, label=sigla)

ax.set_title('Evolução do IDH por estado (1991-2024)')
ax.set_xlabel('Ano')
ax.set_ylabel('IDH')
ax.set_ylim(0.3, 0.9)
ax.legend(ncol=3, bbox_to_anchor=(1.02, 1), loc='upper left', title='UF')
fig.tight_layout()
plt.show()
````
