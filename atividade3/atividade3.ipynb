# Tabela 3 - Pessoas com nível superior completo, por áreas gerais de formação

A tabela possui 3 grandes grupos de áreas:
1. **Total** (todas as áreas / curso superior)
2. **Ciência, Tecnologia, Engenharias e Matemática** (STEM)
3. **Educação, Serviços pessoais, Saúde e Bem-estar**

```python
# Lê o arquivo sem cabeçalho para visualizar a estrutura bruta
bruto = pd.read_excel(path, sheet_name="Sheet1", header=None)
bruto.head(15)
```

```python
# As linhas 3 a 35 contêm os dados (Brasil, regiões e UFs)
# As colunas são:
#   0: região
#   1 a 3: curso superior (Total, Homens, Mulheres)
#   4 a 6: STEM - Ciência, Tecnologia, Engenharias e Matemática
#   7 a 9: Educação, Serviços pessoais, Saúde e Bem-estar
dados = bruto.iloc[3:36].copy()
dados.reset_index(drop=True, inplace=True)
dados
```

```python
# Nomeando as colunas com base na estrutura da tabela
dados.columns = [
    "regiao",
    "sup_total", "sup_homens", "sup_mulheres",
    "stem_total", "stem_homens", "stem_mulheres",
    "ed_total", "ed_homens", "ed_mulheres",
]
dados
```

```python
# Convertendo as colunas numéricas
for col in dados.columns[1:]:
    dados[col] = pd.to_numeric(dados[col], errors="coerce")
dados
```

# Tabela 1 - Homens e mulheres por região com curso superior

```python
superior = dados[["regiao", "sup_homens", "sup_mulheres"]].copy()
superior.columns = ["regiao", "homens", "mulheres"]
superior
```

```python
superior.describe().round(2)
```
