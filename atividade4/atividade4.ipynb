# Atividade 4 - Análise de População por Censo

```python
import pandas as pd
```

## 1. Leitura da tabela

```python
df = pd.read_excel(
    r'CD2022_Populacao_2010_Compatibilizada_20231222.xlsx',
    header=None,
    skiprows=2
)

df.columns = [
    'lixo',
    'UF',
    'COD_UF',
    'COD_MUNIC',
    'NOME_MUNICIPIO',
    'POP_2010_SINOPSE',
    'POP_2010_COMPATIBILIZADA',
    'POP_2022'
]

df = df.drop(columns=['lixo'])

# Converter colunas numéricas
df['POP_2010_SINOPSE'] = pd.to_numeric(df['POP_2010_SINOPSE'], errors='coerce')
df['POP_2010_COMPATIBILIZADA'] = pd.to_numeric(df['POP_2010_COMPATIBILIZADA'], errors='coerce')
df['POP_2022'] = pd.to_numeric(df['POP_2022'], errors='coerce')

print(f'Total de linhas: {len(df)}')
df.head(10)
```

## 2. Tabela de população agregada por estado

```python
pop_estado = (
    df.groupby('UF')[['POP_2010_COMPATIBILIZADA', 'POP_2022']]
    .sum()
    .reset_index()
)

pop_estado['CRESCIMENTO_2010_2022'] = pop_estado['POP_2022'] - pop_estado['POP_2010_COMPATIBILIZADA']

pop_estado = pop_estado.sort_values('CRESCIMENTO_2010_2022', ascending=False)

pop_estado = pop_estado.reset_index(drop=True)

pop_estado
```

## 3. Salvar tabela agregada por estado em CSV

```python
pop_estado.to_csv(r'populacao_por_estado.csv', sep=';', index=False)
print('Arquivo populacao_por_estado.csv salvo com sucesso!')
```

## 4. Tabela por município - ordenada por crescimento populacional

```python
pop_municipio = df[['UF', 'COD_UF', 'COD_MUNIC', 'NOME_MUNICIPIO', 'POP_2010_COMPATIBILIZADA', 'POP_2022']].copy()

pop_municipio['CRESCIMENTO_2010_2022'] = pop_municipio['POP_2022'] - pop_municipio['POP_2010_COMPATIBILIZADA']

pop_municipio = pop_municipio.sort_values('CRESCIMENTO_2010_2022', ascending=False)

pop_municipio = pop_municipio.reset_index(drop=True)

pop_municipio.head(20)
```

## 5. Salvar tabela por município em CSV

```python
pop_municipio.to_csv(r'populacao_por_municipio.csv', sep=';', index=False)
print('Arquivo populacao_por_municipio.csv salvo com sucesso!')
```
